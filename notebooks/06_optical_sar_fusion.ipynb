# Notebook 06: Optical-SAR Fusion

```python
import numpy as np, yaml
from src.models.fusion_model import FusionModel

config = yaml.safe_load(open('configs/colab_8gb.yaml'))
model = FusionModel(config=config)
model.load(config['models']['fusion']['checkpoint'], device='cpu')

optical = np.random.rand(4, 64, 64).astype('float32')
sar = np.random.rand(2, 64, 64).astype('float32')

# Three-way ablation
r_fusion = model.predict(optical, sar, 'Identify land cover.')
r_optical = model.predict_optical_only(optical, 'Optical only.')
r_sar = model.predict_sar_only(sar, 'SAR only.')

print('Fusion:', r_fusion['text'])
print('Optical-only:', r_optical['text'])
print('SAR-only:', r_sar['text'])

from src.evaluation.metrics import multilabel_f1, fusion_ablation
print('\n[Ablation requires real labels from BigEarthNet-MM for valid metrics]')
```
