# Notebook 03: VQA Training and Evaluation

```python
import numpy as np, yaml
from src.models.vqa_model import VQAModel

config = yaml.safe_load(open('configs/colab_8gb.yaml'))
model = VQAModel(config=config)
model.load(config['models']['vqa']['checkpoint'], device='cpu')
print(f'Model loaded: fallback_level={model._fallback_level}')

image = np.random.rand(4, 64, 64).astype('float32')
result = model.predict(image, 'What is visible in this satellite image?')
print('Answer:', result['text'])
print('Confidence:', result['confidence'])
```
