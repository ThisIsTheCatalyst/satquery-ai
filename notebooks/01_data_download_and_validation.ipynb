```python
# Create synthetic dataset for demo
import numpy as np
with open('data/raw/bigearthnet/BigEarthNet.txt', 'w') as f:
    for i in range(100):
        labels = np.random.choice(['Forest', 'Urban fabric', 'Pastures', 'Water bodies'], 2, replace=False)
        f.write(f'patch_{i:04d}\t{chr(44).join(labels)}\n')
print('Synthetic BigEarthNet.txt created (100 patches)')
```
