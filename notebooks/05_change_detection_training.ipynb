# Notebook 05: Change Detection

```python
import numpy as np, yaml
from src.models.change_model import ChangeModel
import matplotlib.pyplot as plt

config = yaml.safe_load(open('configs/colab_8gb.yaml'))
model = ChangeModel(config=config)
model.load(config['models']['change']['checkpoint'], device='cpu')
print(f'Using: {model.backbone} (fallback={model.using_fallback})')

t1 = np.random.rand(3, 64, 64).astype('float32') * 0.5
t2 = t1.copy(); t2[:, 20:40, 20:40] = 0.8

result = model.predict(t1, t2, 'What changed between these dates?')
print(result['text'])

mask = result['spatial_evidence']['mask']
if mask is not None:
    fig, axes = plt.subplots(1,3,figsize=(12,4))
    axes[0].imshow(t1.transpose(1,2,0)); axes[0].set_title('T1')
    axes[1].imshow(t2.transpose(1,2,0)); axes[1].set_title('T2')
    axes[2].imshow(mask, cmap='Reds'); axes[2].set_title('Change Mask')
    for ax in axes: ax.axis('off')
    plt.tight_layout(); plt.show()
```
