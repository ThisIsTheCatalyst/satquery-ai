# Notebook 07: End-to-End Agent Demo

```python
import numpy as np, yaml, json
from src.agent.controller import AgentController

config = yaml.safe_load(open('configs/colab_8gb.yaml'))
controller = AgentController(config, device='cpu')

test_cases = [
    ({'image': np.random.rand(4,64,64).astype('float32')}, 'What is the land cover?'),
    ({'image': np.random.rand(4,64,64).astype('float32')}, 'Highlight the water body.'),
    ({'image_t1': np.random.rand(3,64,64).astype('float32'), 'image_t2': np.random.rand(3,64,64).astype('float32')}, 'What changed?'),
    ({'image_optical': np.random.rand(4,64,64).astype('float32'), 'image_sar': np.random.rand(2,64,64).astype('float32')}, 'Identify land cover.'),
]

for images, query in test_cases:
    result = controller.run(images, query)
    print(f'Task: {result["task"]}')
    print(f'  Answer: {result["text"][:80]}')
    print(f'  Confidence: {result["confidence"]:.2f}  Status: {result["status"]}')
    trace = result['trace']
    print(f'  Latency: {trace.get("latency_seconds", 0)*1000:.0f}ms')
    print()
```
