# Notebook 04: Captioning and Grounding

```python
import numpy as np, yaml
from src.models.grounding_model import GroundingModel, visualize_bbox
import matplotlib.pyplot as plt

config = yaml.safe_load(open('configs/colab_8gb.yaml'))
model = GroundingModel(config=config)
model.load(config['models']['grounding']['checkpoint'], device='cpu')

image = np.random.rand(4, 64, 64).astype('float32')
result = model.predict(image, 'Highlight the water body.')
print('Result:', result['text'])
print('Bbox:', result['spatial_evidence']['bbox'])

if result['spatial_evidence']['bbox']:
    rgb = (image[:3].transpose(1,2,0) * 255).clip(0,255).astype('uint8')
    vis = visualize_bbox(rgb, result['spatial_evidence']['bbox'])
    plt.imshow(vis); plt.title('Grounding'); plt.axis('off'); plt.show()
```
